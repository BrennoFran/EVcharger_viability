```python
# Aplicando Tradução
ctmt['ten_nom_kv'] = ctmt['TEN_NOM'].map(DDA_TENSAO_KV)

# Verifica se algum código ficou sem tradução
nao_mapeados = ctmt[ctmt['ten_nom_kv'].isna()]['TEN_NOM'].unique()
if len(nao_mapeados) > 0:
    print(f"⚠️ Códigos não mapeados: {nao_mapeados}")
    print("Adicione-os ao dicionário DDA_TENSAO_KV e re-execute.")
else:
    print("✓ Todos os códigos foram traduzidos com sucesso.")

print("\nDistribuição de tensões:")
print(ctmt['ten_nom_kv'].value_counts().sort_index())
```

```python
# Carrega SSDMT e calcula comprimento por CTMT
print("Carregando SSDMT (segmentos de média tensão)...")
print("(pode levar 1-3 min em distribuidoras médias)\n")

# Carrega só as colunas necessárias para economizar memória
ssdmt = gpd.read_file(
    GDB_PATH,
    layer="SSDMT",
    columns=['CTMT', 'COMP', 'geometry']
)
print(f"✓ {len(ssdmt):,} segmentos de MT")
print(f"✓ CRS: {ssdmt.crs}")
print(f"\nColunas: {ssdmt.columns.tolist()}")
print(f"\nAmostra:")
display(ssdmt.head())
```

```python
# %% [Célula 5b] — Calcula comprimento total por alimentador
if 'COMP' in ssdmt.columns and ssdmt['COMP'].notna().any():
    print("Usando coluna COMP da BDGD (em metros)")
    ssdmt['_comp_m'] = ssdmt['COMP'].astype(float)
else:
    print("Coluna COMP indisponível — calculando da geometria")
    # Reprojeta para SIRGAS 2000 / Brasil Polyconic (EPSG:5880) — métrico
    ssdmt_metric = ssdmt.to_crs(epsg=5880)
    ssdmt['_comp_m'] = ssdmt_metric.geometry.length

# Agrega por alimentador
comp_por_ctmt = ssdmt.groupby('CTMT').agg(
    comp_total_km=('_comp_m', lambda x: x.sum() / 1000),
    n_segmentos=('_comp_m', 'count')
).reset_index()

print(f"\n✓ {len(comp_por_ctmt):,} alimentadores com geometria")
print("\nDistribuição do comprimento:")
print(comp_por_ctmt['comp_total_km'].describe().round(2))
print("\nAmostra:")
display(comp_por_ctmt.head())
```

```python
# UCBT: a tabela mais pesada — milhões de linhas
print("Carregando UCBT (consumidores de baixa tensão)...")
print("(pode demorar 2-10 min)\n")

# CRÍTICO: carregar SÓ as colunas necessárias
# Se sua versão da BDGD tem nomes diferentes, ajuste aqui
cols_ucbt_essenciais = ['COD_ID', 'CTMT', 'CLAS_SUB']  # mínimo para contagem
cols_ucbt_energia = [f'ENE_{i:02d}' for i in range(1, 13)]  # ENE_01 a ENE_12

# Tenta carregar com energia primeiro
try:
    ucbt = gpd.read_file(
        GDB_PATH, layer="UCBT_tab",  # ⚠️ pode ser "UCBT" sem _tab — ajuste!
        columns=cols_ucbt_essenciais + cols_ucbt_energia
    )
except Exception as e:
    print(f"Tentativa com _tab falhou: {e}")
    print("Tentando UCBT sem sufixo...")
    ucbt = gpd.read_file(
        GDB_PATH, layer="UCBT",
        columns=cols_ucbt_essenciais + cols_ucbt_energia
    )

print(f"✓ {len(ucbt):,} unidades consumidoras BT")
print(f"\nColunas: {ucbt.columns.tolist()}")
print(f"\nAmostra:")
display(ucbt.head())
```

```python
# Conta UCBT por alimentador, removendo "0" (sem associação)
# Filtra zeros e nulos
ucbt_validas = ucbt[
    (ucbt['CTMT'].notna()) &
    (ucbt['CTMT'] != '0') &
    (ucbt['CTMT'] != 0)
].copy()

print(f"UCBT válidas (com CTMT atribuído): {len(ucbt_validas):,}")
print(f"UCBT descartadas (CTMT=0 ou nulo): {len(ucbt) - len(ucbt_validas):,}\n")

# Conta por alimentador
n_ucbt_por_ctmt = ucbt_validas.groupby('CTMT').size().reset_index(name='n_ucbt')

print(f"Alimentadores com UCBT: {len(n_ucbt_por_ctmt):,}")
print(f"\nDistribuição:")
print(n_ucbt_por_ctmt['n_ucbt'].describe().round(0))
print(f"\nAmostra:")
display(n_ucbt_por_ctmt.head())
```

```python
# UCMT (consumidores MT — comércios médios, indústrias)
print("Carregando UCMT...")
try:
    ucmt = gpd.read_file(
        GDB_PATH, layer="UCMT_tab",
        columns=cols_ucbt_essenciais + cols_ucbt_energia
    )
except:
    ucmt = gpd.read_file(
        GDB_PATH, layer="UCMT",
        columns=cols_ucbt_essenciais + cols_ucbt_energia
    )

print(f"✓ {len(ucmt):,} unidades consumidoras MT\n")

ucmt_validas = ucmt[
    (ucmt['CTMT'].notna()) &
    (ucmt['CTMT'] != '0') &
    (ucmt['CTMT'] != 0)
].copy()

n_ucmt_por_ctmt = ucmt_validas.groupby('CTMT').size().reset_index(name='n_ucmt')

print(f"Alimentadores com UCMT: {len(n_ucmt_por_ctmt):,}")
display(n_ucmt_por_ctmt.head())
```

```python
# Calcula demanda média a partir das energias mensais
# Lógica: ENE_XX está em kWh por mês.
# Soma 12 meses → energia anual (kWh/ano)
# Divide por 8760 horas/ano → potência média (kW)

cols_ene_disp_bt = [c for c in cols_ucbt_energia if c in ucbt_validas.columns]
print(f"Colunas de energia BT disponíveis: {cols_ene_disp_bt}")

if cols_ene_disp_bt:
    # Soma anual por UC, depois agrega por alimentador
    ucbt_validas['_ene_anual_kwh'] = ucbt_validas[cols_ene_disp_bt].sum(axis=1)
    ucbt_validas['_dem_med_kw'] = ucbt_validas['_ene_anual_kwh'] / 8760

    dem_bt = ucbt_validas.groupby('CTMT')['_dem_med_kw'].sum().reset_index(
        name='dem_med_ucbt_kw'
    )
else:
    print("⚠️ Sem colunas de energia em UCBT — pulando")
    dem_bt = pd.DataFrame({'CTMT': [], 'dem_med_ucbt_kw': []})

# Mesmo para UCMT
cols_ene_disp_mt = [c for c in cols_ucbt_energia if c in ucmt_validas.columns]
if cols_ene_disp_mt:
    ucmt_validas['_ene_anual_kwh'] = ucmt_validas[cols_ene_disp_mt].sum(axis=1)
    ucmt_validas['_dem_med_kw'] = ucmt_validas['_ene_anual_kwh'] / 8760
    dem_mt = ucmt_validas.groupby('CTMT')['_dem_med_kw'].sum().reset_index(
        name='dem_med_ucmt_kw'
    )
else:
    dem_mt = pd.DataFrame({'CTMT': [], 'dem_med_ucmt_kw': []})

# Junta os dois
demanda = dem_bt.merge(dem_mt, on='CTMT', how='outer').fillna(0)
demanda['dem_total_kw'] = demanda['dem_med_ucbt_kw'] + demanda['dem_med_ucmt_kw']

print(f"\nDemanda calculada para {len(demanda):,} alimentadores")
print("\nEstatísticas (kW médio):")
print(demanda[['dem_med_ucbt_kw', 'dem_med_ucmt_kw', 'dem_total_kw']].describe().round(0))
display(demanda.head())
```

```python
# %% [Célula 9 — VERSÃO CORRIGIDA] Mapeamento de CLAS_SUB alfanumérico

# Tabela DDA-ANEEL: prefixos de CLAS_SUB → classe principal (8 categorias)
# Fonte: Manual de Instruções da BDGD, Anexo I (DDA)
DDA_PREFIXO_CLASSE = {
    'RE': 'residencial',     # RE1, RE2, RE3, RE4, RE5
    'CO': 'comercial',       # CO1 a CO9
    'IN': 'industrial',      # IN
    'RU': 'rural',           # RU1, RU1A, RU1B, RU2, RU3, RU5, RU7, RU8
    'PP': 'poder_publico',   # PP1, PP2, PP3
    'IP': 'iluminacao_publica',  # IP
    'SP': 'servico_publico', # SP1, SP2
    'CP': 'consumo_proprio', # CPR, CPRVE
}

# Verifica primeiro: existe a coluna CLAS_SUB? E como ela está preenchida?
print("Tipo da coluna CLAS_SUB:", ucbt_validas['CLAS_SUB'].dtype)
print("\nPrimeiros 10 valores únicos:")
print(ucbt_validas['CLAS_SUB'].value_counts().head(10))

# Aplica o mapeamento — pega os 2 primeiros caracteres como prefixo
# .astype(str) garante que mesmo se vier numérico ou misto, vire string
ucbt_validas['_prefixo'] = ucbt_validas['CLAS_SUB'].astype(str).str[:2].str.upper()

# Mapeia para a classe principal
ucbt_validas['_classe'] = ucbt_validas['_prefixo'].map(DDA_PREFIXO_CLASSE)

# Sanity check
nao_mapeados = ucbt_validas[ucbt_validas['_classe'].isna()]['CLAS_SUB'].value_counts()
if len(nao_mapeados) > 0:
    print(f"\n⚠️ Códigos NÃO mapeados ({len(nao_mapeados)} tipos):")
    print(nao_mapeados.head(10))
    print("→ Esses caem em 'outros' no pivot")
else:
    print("\n✓ Todos os CLAS_SUB foram mapeados com sucesso")

# Preenche os não mapeados com 'outros'
ucbt_validas['_classe'] = ucbt_validas['_classe'].fillna('outros')

print("\n--- Distribuição final por classe ---")
print(ucbt_validas['_classe'].value_counts())
print(f"\nTotal de UCs: {len(ucbt_validas):,}")
```

```python
# %% [Célula 9b — Pivot: percentuais por classe e por alimentador]

# Pivot: linhas = CTMT, colunas = classe, valores = contagem de UCs
mix = ucbt_validas.pivot_table(
    index='CTMT',
    columns='_classe',
    values='COD_ID',
    aggfunc='count',
    fill_value=0
).reset_index()

# Calcula totais e percentuais
classes_presentes = [c for c in mix.columns if c != 'CTMT']
mix['_total'] = mix[classes_presentes].sum(axis=1)

for c in classes_presentes:
    mix[f'pct_{c}'] = (100 * mix[c] / mix['_total']).round(2)

# Remove a coluna temporária e organiza
mix = mix.drop(columns=['_total'])

# Mostra resumo
print(f"Mix calculado para {len(mix):,} alimentadores")
print(f"\nClasses identificadas: {classes_presentes}\n")

print("Estatísticas dos percentuais:")
cols_pct = [c for c in mix.columns if c.startswith('pct_')]
print(mix[cols_pct].describe().round(1))

print("\nAmostra (primeiras 5 linhas):")
display(mix.head())
```

```python
# %% [Célula 10] — Constrói DataFrame mestre da CTMT enriquecida
print("Construindo DataFrame mestre...\n")

# Começa do CTMT base (já tem ten_nom_kv da Célula 4b)
df = ctmt.copy()

# Renomeia colunas finais para clareza
df = df.rename(columns={
    'COD_ID': 'cod_id',
    'SUB': 'cod_sub',
})

# Junta comprimento
df = df.merge(comp_por_ctmt, left_on='cod_id', right_on='CTMT', how='left')
df = df.drop(columns=['CTMT'])
# Junta UCBT
df = df.merge(n_ucbt_por_ctmt, left_on='cod_id', right_on='CTMT', how='left')
df = df.drop(columns=['CTMT'])

# Junta UCMT
df = df.merge(n_ucmt_por_ctmt, left_on='cod_id', right_on='CTMT', how='left')
df = df.drop(columns=['CTMT'])

# Junta demanda
df = df.merge(demanda, left_on='cod_id', right_on='CTMT', how='left')
df = df.drop(columns=['CTMT'])

# Junta mix
df = df.merge(mix, left_on='cod_id', right_on='CTMT', how='left')
df = df.drop(columns=['CTMT'])

# Anota distribuidora
df['distribuidora'] = NOME_DIST

# Preenche NaN com 0 onde faz sentido
cols_zero = ['n_ucbt', 'n_ucmt', 'dem_med_ucbt_kw', 'dem_med_ucmt_kw',
             'dem_total_kw', 'comp_total_km', 'n_segmentos']
for c in cols_zero:
    if c in df.columns:
        df[c] = df[c].fillna(0)

# Calcula uma feature derivada útil: densidade de carga
df['densidade_kw_por_km'] = np.where(
    df['comp_total_km'] > 0,
    df['dem_total_kw'] / df['comp_total_km'],
    0
)

print(f"✓ DataFrame mestre construído: {len(df):,} alimentadores\n")
print("Colunas finais:")
print(df.columns.tolist())
```

```python
# %% [Célula 11] — Validação
print("=" * 70)
print("VALIDAÇÃO DO DATAFRAME MESTRE")
print("=" * 70)

# 1. Há alimentadores sem dados?
sem_geo = (df['comp_total_km'] == 0).sum()
sem_uc  = (df['n_ucbt'] + df['n_ucmt'] == 0).sum()
sem_dem = (df['dem_total_kw'] == 0).sum()

print(f"\nAlimentadores SEM geometria (km=0):    {sem_geo}/{len(df)}")
print(f"Alimentadores SEM consumidores:         {sem_uc}/{len(df)}")
print(f"Alimentadores SEM demanda calculada:    {sem_dem}/{len(df)}")

# 2. Estatísticas das colunas-chave
print("\n--- Estatísticas das colunas numéricas ---")
cols_num = ['ten_nom_kv', 'comp_total_km', 'n_ucbt', 'n_ucmt',
            'dem_total_kw', 'densidade_kw_por_km']
print(df[cols_num].describe().round(2))

# 3. Top 5 maiores e menores alimentadores por demanda
print("\n--- Top 5 MAIORES alimentadores (por demanda) ---")
display(df.nlargest(5, 'dem_total_kw')[
    ['cod_id', 'cod_sub', 'ten_nom_kv', 'comp_total_km', 'n_ucbt', 'dem_total_kw']
])

print("\n--- Top 5 MENORES alimentadores (por demanda, com dem>0) ---")
display(df[df['dem_total_kw'] > 0].nsmallest(5, 'dem_total_kw')[
    ['cod_id', 'cod_sub', 'ten_nom_kv', 'comp_total_km', 'n_ucbt', 'dem_total_kw']
])
```
